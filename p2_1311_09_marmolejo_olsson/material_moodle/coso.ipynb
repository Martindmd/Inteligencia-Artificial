```
from sklearn.model_selection import train_test_split
```

```
df_train, df_test = train_test_split(df, test_size=0.3, random_state=42,
                                     stratify=df["target"]
                                     )
```

```
df.shape
```

```
df_train.shape
```

```
df_test.shape
```

```
100*df["target"].value_counts()/len(df)
```

```
100*df_test["target"].value_counts()/len(df_test)
```

```
X_train = df_train.drop("target", axis=1) # entrada a mi modelo
y_train = df_train["target"] # salida deseada de mi modelo

X_test = df_test.drop("target", axis=1)
y_test = df_test["target"]
```

```
from sklearn.naive_bayes import GaussianNB
```

```
model1 = GaussianNB()
model1.fit(X_train, y_train)
```

```
model1.predict(X_train).shape
```

```
X_train.shape
```

```
100*np.mean(model1.predict(X_train) == y_train)
```

```
100*model1.score(X_train, y_train)
```

```
100*np.mean(model1.predict(X_test) == y_test)
```

```
100*model1.score(X_test, y_test)
```

```
from sklearn.neighbors import KNeighborsClassifier
```

```
model2 = KNeighborsClassifier()
model2.fit(X_train, y_train)
```

```
100*model2.score(X_train, y_train)
```

```
100*model2.score(X_test, y_test)
```

```
X_train.describe().T[["mean", "std"]]
```

```
# Estandarización (forma 1):

X_train_std = (X_train - X_train.mean())/X_train.std()
X_test_std  = (X_test - X_train.mean())/X_train.std()
```

```
X_train.mean()
```

```
model3 = KNeighborsClassifier()
model3.fit(X_train_std, y_train)
```

```
100*model3.score(X_train_std, y_train)
```

```
100*model3.score(X_test_std, y_test)
```
